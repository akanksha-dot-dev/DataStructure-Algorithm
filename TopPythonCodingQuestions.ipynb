```python
#Palindrome String
#A. By Using While Loop

def palindrome(s):
    n=len(s)
    first=0
    last=n-1
    while(first<last):
        if s[first] == s[last]:
            first += 1 #first=first+1
            last -= 1 #last=last-1
        else:
            return False
    return True

s="nitind"
palindrome(s)
print(palindrome(s))
```

```python
#Palindrome Number
#In Case of Numbers
#By Using While Loop
def palindrome(n):
    temp = n
    rev_n=0
    while(temp>0):
        digit=temp%10
        #print(digit)
        rev_n=rev_n*10 + digit
        #rev_n=0*10 +6 =6
        temp = temp//10
    if n==rev_n:
        return True
    else:
        return False

n=12345654321
print(palindrome(n))
```

Fibonacci Series
0,1,1,2,3,5,8,13....

```python
#Fibonacci Series
#Find the Fibonacci Series from 0 to nth number using the while loop.
# 0 1 1 2 3 5 8 13 21
# By using the While Loop

def fibonacci(n):
    a,b=0,1
    print(a)
    while (b<n):
        print(b)
        c=a+b
        a=b
        b=c
fibonacci(50)
```

```python
#Fibonacci Series
# By using the While Loop without using 3rd Variable.

def fibonacci(n):
    a,b=0,1
    print(a)
    while (b<n):
        print(b)
        a,b=b,a+b
fibonacci(50)
```

```python
#By using For Loop
def fibonnaci2(n):
    a,b=0,1
    if n==1:
        print(a)
    else:
        print(a)
        print(b)
        for i in range(2,n):
            c=a+b
            a=b
            b=c
            print(c)
fibonnaci2(10)
```

```python
#By using Recurssion
#Best Approach For Fibonacci Series
def fibonacci3(n):
    if n<=1:
        return n
    else:
        return(fibonacci3(n-1) + fibonacci3(n-2))
n=10
if n<=0:
    print("Invalid")
else:
    for i in range(n):
        print(fibonacci3(i))   
```

```python
#Compress String
#AABBCC => A2B2C2
#Input = AABBCC
#Output = A2B2C2
#Best Approach For Compress String Format
#By Using the For Loop
def compress(s):
    n = len(s)
    new_s = ''
    count = 1
    for i in range(n-1):
        if s[i] == s[i+1]:
            count += 1
        else:
            new_s=new_s + s[i] +str(count)
            #new_s += (s[i]+str(count))
            count = 1
    new_s = new_s + s[n-1] + str(count)
    return new_s

s = 'aabbcccaaaaaffffeeeeiiii'
print(compress(s))
```

```python
#Compress String
#AABBCC => A2B2C2
#Input = AABBCC
#Output = A2B2C2

#By Using the While Loop
def compress1(s):
    n=len(s)
    i=0
    new_s=''
    while (i<n-1):
        count = 1
        while (i<n-1 and s[i] == s[i+1]):
            count += 1
            i+=1
        i+=1
        new_s = new_s + s[i-1] + str(count)
    return new_s
s = 'aabbcccaaaaaffffeeeeiiii'
print(compress1(s))
```

```python
#Best Approach for FizzBuzz Problem
#FizzBuzz
#3* => Fizz
#5* => Buzz
#15* => FizzBuzz
#dict = {3:'Fizz', 5:'Buzz'}

#FizzBuzz Problems
#If Number divisible by 3 => Print Fizz
#If Number divisible by 5 => Print Buzz
#If Number divisible by 15 => Print FizzBuzz
#Else Print the number

# 1 - 1
# 2 - 2
# 3 - Fizz
# 4 - 4
# 5 - Buzz
# 6 - Fizz
#..............
# 15 - FizzBuzz

#By using For Loop

def fizzbuzz(n):
    for i in range(1,n+1):
        if i%3 == 0 and i%5 == 0: #i%15 == 0
            print("FizzBuzz")
        elif i%3 == 0:
            print("Fizz")
        elif i%5 == 0:
            print("Buzz")
        else:
            print(i)
fizzbuzz(20)
```

```python
#FizzBuzz using Dictionary

def fizzbuzz1(n):
    d = {3: 'Fizz',
         5: 'Buzz'}
    for i in range(1,n+1):
        result = ''
        for k, v in d.items():
            if i%k == 0:
                result += v
        if not result:
            result = i
        print(result)
fizzbuzz1(15)
```
