```python
#Palindrome for String
#By Using the Split Method
def palindrome(s):
    temp = s[::-1]
    if temp == s:
        print("The given string is a Palindrome String.")
    else:
        print("The given string is not a Palindrome String.")

s="nitin"
palindrome(s)
print(palindrome(s))  
```

```python
#Palindrome for String
#By Using Indexing/ using the For Loop

def palindrome2(s):
    n = len(s)
    for i in range(n):
        if s[i] != s[n-i-1]:
            return False
    return True
    
s="nitin"
print(palindrome2(s))
```

```python
#Palindrome for String
#By Using Inbuilt Functions - reversed and Join

def palindrome3(s):
    reverse_s = reversed(s)
    for i in reverse_s:
        print(i)
    print(reverse_s)
    temp = ''.join(reversed(s))
    print(temp)
    if s == temp:
        print("The given string is a Palindrome String.")
    else:
        print("The given string is not a Palindrome String")
palindrome3("nitin")
```

```python
#Palindrome for String
#By Using the While Loop

def palindrome(s):
    n = len(s)
    first = 0
    last = n-1
    while(first<last):
        if s[first] == s[last]:
            first += 1
            last -= 1
        else:
            return False
    return True

s = "nitin"
print(palindrome(s))
```

```python
#Palindrome Number
#In case of Palindrome Numbers
#By using Split Method - Converting Numbers to String
def palindromeNum(n):
    s = str(n)
    temp = s[::-1]
    if s == temp:
        print("The given number is a Palindrome Number.")
    else:
        print("The given number is not a Palindrome Number.")

n = 23432
palindromeNum(n)
```

```python
#Palindrome Number
#By Using While Loop
def paliNum(n):
    temp = n
    rev_n=0
    while(temp>0):
        digit = temp % 10
        rev_n = rev_n * 10 + digit
        temp = temp // 10
    if n == rev_n:
        print("Yes, The given number is Palindrome Number.")
    else:
        print("No, The given number is not Palindrome Number.")

n=7893987
paliNum(n)
```

```python
#Fibonnaci Series
#Find the Fibonnaci series from 0 to nth number using while loop
#0,1,1,2,3,5,8,13........
#By Using While Loop
def fibonnaci(n):
    a,b=0,1
    print(a)
    while(b<n):
        print(b)
        c = a+b
        a = b
        b = c
fibonnaci(9)
```

```python
#Fibonnaci Series
#Find the Fibonnaci series from 0 to nth number using while loop
#0,1,1,2,3,5,8,13........
#By Using While Loop, without using any 3rd Variable
def fibo(n):
    a,b=0,1
    print(a)
    while(b<n):
        print(b)
        a,b=b,a+b

fibo(9)
```

```python
#Fibonaaci Series 
#By Using the For Loop

def forfibo(n):
    a,b=0,1
    print(a)
    print(b)
    for i in range(n):
        a,b=b,a+b
        print(i)

forfibo(5)
```
