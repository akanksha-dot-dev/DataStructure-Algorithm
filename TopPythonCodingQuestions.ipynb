Palindrome String / Palindrome Number

```python
#Palindrome String
#A. By Using Split Method

def palindrome(s):
    temp=s[::-1]
    if s == temp:
        print("Yes!! It's a Palindrome Number.")
    else:
        print("No!! It is not a Palindrome Number.")
s="ruhur"
palindrome(s)
```

```python
#Palindrome String
#A. By Using Indexing / By Using For Loop

def palindrome(s):
    n=len(s)
    for i in range(n):
        if s[i] != s[n-i-1]:
            return False
            #print("It is not Palindrome.")
    return True
    #print("It is Palindrome.")
s = "riverevir"
print(palindrome(s))
```

```python
#Palindrome String
#A. By Using While Loop

def palindrome(s):
    n=len(s)
    first=0
    last=n-1
    while(first<last):
        if s[first] == s[last]:
            first += 1 #first=first+1
            last -= 1 #last=last-1
        else:
            return False
    return True

s="nitind"
palindrome(s)
print(palindrome(s))
```

```python
#Palindrome Number
#In Case of Numbers
#By Using Split Method - Converting Number to String
#Not Preferred Case
def palindrome(n):
    s=str(n)
    temp=s[::-1]
    if s == temp:
        return True
    else:
        return False
    

n = 12321
print(palindrome(n))
```
