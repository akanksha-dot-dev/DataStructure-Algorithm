Types of Data Structures

1. Primitive Data Structures => In Python, Primitive Data Structures are basic data types that can not be broken into simpler data types. Primitive Data Structures are of fixed size and they are usually smaller in size. Primitive Data Structures are Simple. Primitive Data Structures are used for Simple Operations. Primitive Data Types are presented in memory as a simple data type.

Main Primitive Datatypes in Python are:
1. Integer
Float
String
Boolean
