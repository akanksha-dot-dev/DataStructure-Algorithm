Palindrome String / Palindrome Number

```python
#Palindrome String
#A. By Using Split Method

def palindrome(s):
    temp=s[::-1]
    if s == temp:
        print("Yes!! It's a Palindrome Number.")
    else:
        print("No!! It is not a Palindrome Number.")
s="ruhur"
palindrome(s)
```

```python
#Palindrome String
#A. By Using Indexing / By Using For Loop

def palindrome(s):
    n=len(s)
    for i in range(n):
        if s[i] != s[n-i-1]:
            return False
            #print("It is not Palindrome.")
    return True
    #print("It is Palindrome.")
s = "riverevir"
print(palindrome(s))
```

```python
#Palindrome String
#A. By Using While Loop

def palindrome(s):
    n=len(s)
    first=0
    last=n-1
    while(first<last):
        if s[first] == s[last]:
            first += 1 #first=first+1
            last -= 1 #last=last-1
        else:
            return False
    return True

s="nitind"
palindrome(s)
print(palindrome(s))
```

```python
#Palindrome Number
#In Case of Numbers
#By Using While Loop
def palindrome(n):
    temp = n
    rev_n=0
    while(temp>0):
        digit=temp%10
        #print(digit)
        rev_n=rev_n*10 + digit
        #rev_n=0*10 +6 =6
        temp = temp//10
    if n==rev_n:
        return True
    else:
        return False

n=12345654321
print(palindrome(n))
```

Fibonacci Series
0,1,1,2,3,5,8,13....

```python
#Fibonacci Series
#Find the Fibonacci Series from 0 to nth number using the while loop.
# 0 1 1 2 3 5 8 13 21
# By using the While Loop

def fibonacci(n):
    a,b=0,1
    print(a)
    while (b<n):
        print(b)
        c=a+b
        a=b
        b=c
fibonacci(50)
```

```python
#Fibonacci Series
# By using the While Loop without using 3rd Variable.

def fibonacci(n):
    a,b=0,1
    print(a)
    while (b<n):
        print(b)
        a,b=b,a+b
fibonacci(50)
```
