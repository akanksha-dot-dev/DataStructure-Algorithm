Types of Algorithm :

1. Sorting Algorithm : To sort data in ascending or descending order.e.g. Bubble Sort, Insertion Sort, Selection Sort, etc.

2. Searching Algorithm: To Find a specific value in a data set. e.g. Linear Search, Binary search etc.

What is Big O?
Big O is the language and metric we use to describe the efficiency of algorithms.

1. Time Complexity: A way of showing how the runtime of a function increases as the size of the input increases.

2. Space Complexity: Space Complexity is the amount of memory that a code use.

3. Types of Runtimes: O(N), O(N2), O(2N)
Time Complexity: O(wh)

4. Big O Notations: 
    1. Big O: It is a complexity that is going to be less or equal to the worst case.
    2. Big-Omega: It is a complexity that is going to be at least more than the best case.
    3. Big-Theta: It is a complexity that is within bounds of the worst and the best cases.

O(n) Time Complexity (Linear Time Complexity) 

O(n) Time Complexity will grow in the direct proportion to the size of the input data. For example, the below program is performing the operations n times if the value of the input is n. The number of operations increases if we increases the value of the input.As the input grows the number of operations performing also grows with time.

```python
def print_items(n):
    for i in range(n):
        print(i)

print_items(2)
```

Drop Constants (Remove Constants)




```python
def print_items(n):
    for i in range(n):
        print(i)
    for j in range(n):
        print(j)

print_items(2)
```

Quadratic Time Complexity O(n^2)



```python
def print_items(n):
    #O(N)
    for i in range(n): #O(N)
        for j in range(n): #O(N)
            print(i,j)
print_items(2)
```

Drop Non Dominant Terms



```python
def print_items(n):
    #O(N)
    for i in range(n): 
        for j in range(n): #O(N^2)
            print(i,j)
    for k in range(n):#O(N)
        print(k)
print_items(2)
```

Space Complexity 

Space Complexity is the measure of the amount of working storage that an algorithm need. i.e. how much memory in the worst case is needed at any point in the algorithm. How the sapce needed grows as the size of input increases.

```python
def sum(n):
    if n<=0:
        return 0
    return n + sum(n-1)

sum(3)
```

For Array Traversal, 
1. Time Complexity : O(n)
2. Space Complexity : O(1)




```python
from array import *

arr1= array('i',[1,2,3,4,5,6])
arr2= array('d', [1.2, 1.5, 1.6])

arr1.insert(2,9)
print(arr1)

def traverseArray(array): #---------O(n)
    for i in array: #---------O(n)
        print(i)  #---------O(1)

traverseArray(arr1)
traverseArray(arr2)
```

Accessing an element of Array



```python
#Access array element=> 
# Time Complexity = O(1) 
# Space Complexity = O(1)

from array import *

arr1 = array('i',[1,2,3,4,5,6])

def accessElement(array, index):
    if index >= len(array): #---O(1)
        print('There is no any element in this index.') #--O(1)
    else:
        print(array[index]) #---O(1)

accessElement(arr1,9)
#print(accessElement(arr1,3))
```

Searching for an element in Array => Linear Search Algorithm

1. Time Complexity for Len Function and Range Function
    1. Len Function: Time Complexity of Len Function is O(1), as it is a constant time operation that retrieves the stored len value of Array Object. It does not impact overall Time Complexity of the Linear Search at all.
    
    2. Range Function : Range Function is used to crate range object representing a sequence of integers from 0 upto but not including the len of input array. The Time Complexity for creating a Range Object is O(1) as it does not generates the actual sequence of integers but rather creates an iterator that can be used to produce the numbers on demand.
    
    The Space Complexity for both the fuction is O(1) because here a additional memory is not required while performing any operation.


```python
import array
#Time Complexity for the Linear Search is O(n)
#Time Complexity of len Function is O(1)
#Time Complexity of range Function is O(1)
#Space Complexity of both function is O(1) as there is no any extra space is required at all.
my_array1 = array.array('i',[1,2,3,4,5,6])

def linear_search(arr, target):
    for i in range(len(arr)): #---O(n)
        if arr[i] == target: #---O(1)
            return i
    return -1

print(linear_search(my_array1, 8))
```

Time and Space Complexity of Array



1. Creating an Empty Array
    When we are creating an empty array in python using array module or numpy module,the Time Complexity = O(1). Because in this case, space in the memory is not allocated to array. Hence, Space Complexity = O(1). 
2. Creating an array with elements
    Now, when we are creating an array with the element n, then the Time Complexity = O(N). Because here, we have to loop through a sequence that is provided as the parameter and insert them one by one in the array. Space Complexity =O(N), Becuase here N numbers of contegious cells would be allocated in a memory. And, N is the number of elements. Hence, Space Complexity=O(N).
3. Inserting a value in an Array
    When we are inserting an element in the array, then in worst case the Time Complexity would be O(N). Because,if we insert element at the begining of the array, of in the middle of the array, All elements have to shift one space right.That is why, the Time Complexity = O(N). Space Complexity = O(1), because in this case, in the memory only one cell memory will be allocated. Hence, Space Complexity = O(1).
4. Traversing a given array
    When we are traversing an array, then we have to visit each element of the array. Time Complexity = O(N). Space Complexity = O(1) as the additional space memory is not required here during traversing the array.
5. Accessing a given Cell:
    When it comes to accessing the given value in the array.Time Complexity = O(1) and Space Complexity = O(1). Because, we know that we are accessing the element of the array based on index. And, this feature makes the array most used data structure.
6. Searching a given value
    Now, When we come to searching the given value in the array.Time Complexity = O(N) Because here, we have to look through all the elemnets of the array, and check that if this is the element that we are searching or not. In this case it is going to be linear search, Hence, Time Complexity = O(N).Space Complexiyt is O(1) as additional memory is not required here.
7. Deleting a given value
    When we are deleting the element from the array, all the rest elements will be shifted left because we can not leave the empty space in the middle or the begining of the array as it is a contegious memory. Hence, Time Complexity = O(N) and Space Complexity = O(1) as extra space is not required at all for this operation.
