```python
#Palindrome String
#A. By Using While Loop

def palindrome(s):
    n=len(s)
    first=0
    last=n-1
    while(first<last):
        if s[first] == s[last]:
            first += 1 #first=first+1
            last -= 1 #last=last-1
        else:
            return False
    return True

s="nitind"
palindrome(s)
print(palindrome(s))
```

Fibonacci Series
0,1,1,2,3,5,8,13....

```python
#Fibonacci Series
#Find the Fibonacci Series from 0 to nth number using the while loop.
# 0 1 1 2 3 5 8 13 21
# By using the While Loop

def fibonacci(n):
    a,b=0,1
    print(a)
    while (b<n):
        print(b)
        c=a+b
        a=b
        b=c
fibonacci(50)
```

```python
#Fibonacci Series
# By using the While Loop without using 3rd Variable.

def fibonacci(n):
    a,b=0,1
    print(a)
    while (b<n):
        print(b)
        a,b=b,a+b
fibonacci(50)
```

```python
#Compress String
#AABBCC => A2B2C2
#Input = AABBCC
#Output = A2B2C2
#Best Approach For Compress String Format
#By Using the For Loop
def compress(s):
    n = len(s)
    new_s = ''
    count = 1
    for i in range(n-1):
        if s[i] == s[i+1]:
            count += 1
        else:
            new_s=new_s + s[i] +str(count)
            #new_s += (s[i]+str(count))
            count = 1
    new_s = new_s + s[n-1] + str(count)
    return new_s

s = 'aabbcccaaaaaffffeeeeiiii'
print(compress(s))
```

```python
#Best Approach for FizzBuzz Problem
#FizzBuzz
#3* => Fizz
#5* => Buzz
#15* => FizzBuzz
#dict = {3:'Fizz', 5:'Buzz'}

#FizzBuzz Problems
#If Number divisible by 3 => Print Fizz
#If Number divisible by 5 => Print Buzz
#If Number divisible by 15 => Print FizzBuzz
#Else Print the number

# 1 - 1
# 2 - 2
# 3 - Fizz
# 4 - 4
# 5 - Buzz
# 6 - Fizz
#..............
# 15 - FizzBuzz

#By using For Loop

def fizzbuzz(n):
    for i in range(1,n+1):
        if i%3 == 0 and i%5 == 0: #i%15 == 0
            print("FizzBuzz")
        elif i%3 == 0:
            print("Fizz")
        elif i%5 == 0:
            print("Buzz")
        else:
            print(i)
fizzbuzz(20)
```

```python
#FizzBuzz using Dictionary

def fizzbuzz1(n):
    d = {3: 'Fizz',
         5: 'Buzz'}
    for i in range(1,n+1):
        result = ''
        for k, v in d.items():
            if i%k == 0:
                result += v
        if not result:
            result = i
        print(result)
fizzbuzz1(15)
```

What is Decorator? Explain with Example.

A Decorator is just a function that takes another function as an argument, add some kind of functionality and then returns another function.
All of this without altering the source code of the original function that you passed in.

```python
def decorator_func(func):
    def wrapper_func():
        print("wrapper_func Worked")
        return func()
    print("decorater_func Worked")
    return wrapper_func

def show():
    print("Show Worked")
decorator_show = decorator_func(show)
decorator_show()
```

```python
#Alternative of the above Program for the Decorator

@decorator_func
def display():
    print('display worked')
display()
```

Difference between List Comprehension and Dict Comprehension.
List Comprehension Syntax : [expression for item in iterable if conditional]
Dict Comprehension Syntax : {key:value for (key,value) in iterable if conditional}

```python
#Before List Comprehension (Common Way)

l=[]
for i in range(10):
    if i%2:
        l.append(i)
print(i)
print(l)

#Using List Comprehension
ls = [i for i in range(10) if i%2]
print(ls)
```

Difference between the Generator and Iterator?

Generator : Generators are iterators which can execute only once.
Generator uses "yield" keyword.
Generators are mostly used in loops to generate an iterator by returning all the values in the loop without affecting the iteration of the loop.
Every Generator is an Iterator.

Iterator: An Iterator is an object which contains a countable number of values and it is used to iterate over iterable objects like list, tuple, sets, etc.
Iteroators are used mostly to iterate or convert other objects to an iterator using the iter() function.
Iterator uses the iter() and next() function.
Every Iterator is not a generator.
