```python
#Palindrome for String
#By Using the Split Method
def palindrome(s):
    temp = s[::-1]
    if temp == s:
        print("The given string is a Palindrome String.")
    else:
        print("The given string is not a Palindrome String.")

s="nitin"
palindrome(s)
print(palindrome(s))  
```

```python
#Palindrome for String
#By Using Indexing/ using the For Loop

def palindrome2(s):
    n = len(s)
    for i in range(n):
        if s[i] != s[n-i-1]:
            return False
    return True
    
s="nitin"
print(palindrome2(s))
```

```python
#Palindrome for String
#By Using Inbuilt Functions - reversed and Join

def palindrome3(s):
    reverse_s = reversed(s)
    for i in reverse_s:
        print(i)
    print(reverse_s)
    temp = ''.join(reversed(s))
    print(temp)
    if s == temp:
        print("The given string is a Palindrome String.")
    else:
        print("The given string is not a Palindrome String")
palindrome3("nitin")
```
